```python
%sql
 CREATE TABLE training.delta_demo.people (first_name STRING, age LONG) USING delta TBLPROPERTIES (delta.enableChangeDataFeed = true)
```

```python
df = spark.createDataFrame([("Bob", 23), ("Sue", 25), ("Jim", 27)]).toDF("first_name", "age")
df.write.mode("append").format("delta").saveAsTable("training.delta_demo.people")
display(df)
```

```python
%sql
select * from training.delta_demo.people
```

```python
%sql
delete from training.delta_demo.people where first_name = 'Bob'
```

```python
%sql
update training.delta_demo.people set age=30 where first_name = 'Jim'
```

```python
df=spark.read.format("delta")\
    .table("training.delta_demo.people")
display(df)
```

```python
df=spark.read.format("delta")\
    .option("readChangeFeed", "true")\
    .option("startingVersion", 0)\
    .table("training.delta_demo.people")
display(df)
```
